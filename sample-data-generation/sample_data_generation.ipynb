```python
#Prepare mask for each instructor, the number of lectures per day does not exceed 2
nb_lec_02 = []
nb_lec_01 = []
for mask in range(16): 
    cnt_bits = 0
    for i in range(4): 
        cnt_bits += ((mask >> i) & 1)
    if cnt_bits == 2: 
        nb_lec_02.append(mask)
    if cnt_bits == 1: 
        nb_lec_01.append(mask)

print(nb_lec_01)
print(nb_lec_02)
rand_mask = nb_lec_01
rand_mask.extend(nb_lec_02)
rand_mask.append(0)
print(rand_mask)
```

```python

file_content = f"Assignment ID,Course ID,Instructor ID,Shift\n"
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat']
for i in range(len(instructor_names)): 
    instructor_id = f"ins{i:05d}"
    expertise = instructor_names[i][1]
    assigned_courses = []
    for (j, course) in enumerate(course_names): 
        if course[1] == expertise: 
            assigned_courses.append(j)
    for day in range(6): 
        mask = random.choice(rand_mask)
        for s in range(4): 
            if ((mask >> s) & 1) == 1: 
                assignment_matrix[i][day][s] = random.choice(assigned_courses)

cnt_assignments = 0
asg_ids = []
for i in range(len(instructor_names)): 
    instructor_id = f"ins{i:05d}"
    for day in range(6): 
        for s in range(4):
                if assignment_matrix[i][day][s] != "Empty":
                    stt = assignment_matrix[i][day][s]
                    course_id = f"crs{stt:05d}"
                    file_content += f"asg{cnt_assignments:07d},{course_id},{instructor_id},{days[day]}-{s}\n"
                    asg_ids.append(f"asg{cnt_assignments:07d}")
                    cnt_assignments += 1

                    
with open("course_instructor.csv", "w") as f: 
    f.write(file_content)
```

```python
import random
from datetime import datetime, timedelta

def get_random_date(start_date_str="2026-04-30", end_date_str="2026-05-05"):
    # chuyển đổi string sang đối tượng datetime
    start = datetime.strptime(start_date_str, "%Y-%m-%d")
    end = datetime.strptime(end_date_str, "%Y-%m-%d")
    
    # tính số ngày chênh lệch
    delta = end - start
    int_delta = delta.days
    
    # lấy ngẫu nhiên một số ngày trong khoảng chênh lệch
    random_days = random.randrange(int_delta + 1)
    
    # cộng vào ngày bắt đầu
    return str((start + timedelta(days=random_days)).date())
```

```python
from datetime import datetime

def sort_list_by_date(data_list, date_format='%Y-%m-%d'):
    """
    Hàm sắp xếp list các dictionary theo ngày từ cũ đến mới.
    :param data_list: List đầu vào (dạng list of dicts)
    :param key_name: Tên của key chứa ngày (mặc định là 'date')
    :param date_format: Định dạng ngày tháng trong chuỗi (mặc định YYYY-MM-DD)
    """
    return sorted(
        data_list, 
        key=lambda x: datetime.strptime(x[2], date_format)
    )
```

```python
import pandas as pd
import re

def standardize_course_id(file_path, output_path, id_column_index=1):
    # 1. Đọc file
    df = pd.read_csv(file_path)
    
    # 2. Hàm xử lý chuỗi: Tách phần chữ và phần số, sau đó thêm số 0
    def pad_id(original_id):
        if pd.isna(original_id):
            return original_id
        
        # Tìm phần số trong chuỗi (ví dụ từ crs01 lấy ra 01)
        match = re.search(r'\d+', str(original_id))
        if match:
            number_part = match.group()
            # Thêm số 0 vào trước cho đủ 5 chữ số
            padded_number = number_part.zfill(5)
            # Ghép lại với tiền tố 'crs'
            return f"crs{padded_number}"
        return original_id

    # 3. Áp dụng cho cột CourseID (thường là cột thứ 2 - index 1 trong file lectures)
    col_name = df.columns[id_column_index]
    df[col_name] = df[col_name].apply(pad_id)
    
    # 4. Lưu lại file đã sạch
    df.to_csv(output_path, index=False)
    print(f"✅ Đã chuẩn hóa CourseID trong file {file_path} và lưu vào {output_path}")

# --- THỰC THI ---
# Chuẩn hóa file bài giảng
standardize_course_id('lectures.csv', 'lectures.csv', id_column_index=1)
```
